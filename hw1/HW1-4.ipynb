4. Дан 1D массив, удалите все элементы которые входят в диапазон `[3, 8]` inplace

````python
array = np.array([0, 4, 4, 2, 1, 4, 8, 7, 10, 15])
mask = (array < 3) | (array > 8)
numbers_count = mask.sum()
array[:numbers_count] = array[mask]
array.resize(numbers_count, refcheck=False)

print(array)
````

8. Как конвертировать массив float(32 bit) к массиву целых чисел (integer 32 bit) in place?

### 2. Pandas

Найдите ответы на вопросы или выполните действия с предложенным датасетом

````python
import pandas as pd
````

````python
# Load dataset here
df = pd.read_csv('russian-housing-market.csv')

df
````

1. Посчитайте количество записей в датасете с отсутствующей информацией о жилой площади жилого помещения

````python
nan_life_sq_count = df['life_sq'].isna().sum()

print(nan_life_sq_count)
````

2. Предложите способ заменить отсутствующие значения из пункта 1 и реализуйте его. Способ нужно аргументировать(сказать почему это достаточно хорошее приближение)

````python
# Просто беру медиану, т.к. она даёт хорошую оценку, устойчивую к выбросам.
df['life_sq'] = df['life_sq'].fillna(df['life_sq'].median())
````

3. Какие значения имеет 25, 50 и 75 перцентили полной площади помещения

````python
quantiles = df['full_sq'].quantile([0.25, 0.5, 0.75])

print(quantiles)
````

4. Посчитайте среднюю стоимость продажи помещений с общей площадью от 50 до 70 квадратных метров

````python
mask = (df['full_sq'] >= 50) & (df['full_sq'] <= 70)
mean_price = df[mask]['price_doc'].mean()

print(mean_price)
````

5. Какая максимальная этажность здания представлена в датасете

````python
max_floor = df['max_floor'].max()

print(max_floor)
````

6. Выявите район с самым дорогим жильем

````python
# Район с самым дорогим жильём - район, в котором самая большая цена жилья в среднем
rich_sub_area = df.groupby('sub_area')['price_doc'].mean().idxmax()

print(rich_sub_area)

# Для альтернативной трактовки района с самым дорогим жильём
# df.groupby('sub_area')['price_doc'].max().idxmax()
````

7. Посчитайте среднюю стоимость за квадратный метр жилья в Митино(Mitino)

````python
df_mitino = df[df['sub_area'] == 'Mitino']
mean_price_per_full_sq = (df_mitino['price_doc'] / df_mitino['full_sq']).mean()

print(mean_price_per_full_sq)
````

8. Посчитайте среднюю площадь комнаты в Бибирево

````python
df_bibirevo = df[df['sub_area'] == 'Bibirevo']
mean_room_sq = (df_bibirevo['life_sq'] / df_bibirevo['num_room']).mean()

print(mean_room_sq)
````

9. Проанализируйте есть ли разница в цене между квартирами на 1 и 9 этажах

````python
# Т.е., насколько средняя стоимость квартир на 1 этаже выше средней стоимости квартир на 9 этаже
mean_first_floor_price = df[df['floor'] == 1]['price_doc'].mean()
mean_nine_floor_price = df[df['floor'] == 9]['price_doc'].mean()
difference = mean_first_floor_price - mean_nine_floor_price

print(difference)
````

10. Найдите минимальную стоимость квартиры с жилой площадью не менее 60 квадратных метров, не менее чем 3мя комнатами и не ниже чем на 5 этаже

````python
mask = (df['life_sq'] >= 60) & (df['num_room'] >= 3) & (df['floor'] >= 5)
min_price = df[mask]['price_doc'].min()

print(min_price)
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import plotly.express as px
import plotly.io as pio


pio.renderers.default = 'colab'

df['timestamp'] = pd.to_datetime(df['timestamp'])
````

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinski Zaton) с течением времени

````python
prices = (
    df.loc[df['sub_area'] == 'Nagatinskij Zaton']
    .set_index('timestamp')['price_doc']
    .resample('MS').mean()
)

px.line(
    x=prices.index, y=prices, markers=True,
    title='Средняя цена в Нагатинском Затоне',
    labels={'x': 'Месяц', 'y': 'Средняя цена, руб.'},
).show()
````

2. Постройте гистограмму распределения цен на квартиры с общей площадью более 50 кв.метров

````python
prices = df.loc[df['full_sq'] > 50, 'price_doc']

px.histogram(
    x=prices, nbins=50,
    title='Цены квартир площадью более 50 м^2',
    labels={'x': 'Цена, руб.'},
).update_layout(yaxis_title='Количество квартир').show()
````

3. Визуализируйте количество проданных квартир в разных районах

````python
counts = df['sub_area'].value_counts().sort_values()

px.bar(
    x=counts, y=counts.index, orientation='h',
    height=20 * len(counts),
    title='Количество продаж по районам',
    labels={'x': 'Количество квартир', 'y': 'Район'},
).show()
````

4. Постройте box plot отображающий распределение полных площадей квартир в Басманном районе

````python
areas = df.loc[df['sub_area'] == 'Basmannoe', 'full_sq']

px.box(
    x=areas, height=280,
    title='Полная площадь квартир в Басманном районе',
    labels={'x': 'Площадь, м^2'},
).show()
````

5. Визуализируйте зависимость средней стоимости квартир в зависимости от этажа

````python
prices = df.groupby('floor')['price_doc'].mean()

px.bar(
    x=prices.index, y=prices,
    title='Средняя цена квартиры по этажам',
    labels={'x': 'Этаж', 'y': 'Средняя цена, руб.'},
).show()
````

6. Покажите как меняется средняя площадь квартир в зависимости от района

````python
areas = df.groupby('sub_area')['full_sq'].mean().sort_values()

px.bar(
    x=areas, y=areas.index, orientation='h',
    height=20 * len(areas),
    title='Средняя площадь квартир по районам',
    labels={'x': 'Средняя площадь, м^2', 'y': 'Район'},
).show()
````

7. Постройте график изменения медианной цены на квартиру менее 50 кв. метров за период с 2011 до 2014 года*

\* Задание на доп. балл

````python
mask = (
    (df['full_sq'] < 50)
    & (('2011-01-01' <= df['timestamp']) & (df['timestamp'] < '2015-01-01'))
)
prices = (
    df.loc[mask].set_index('timestamp')['price_doc']
    .resample('MS').median()
)

px.line(
    x=prices.index, y=prices, markers=True,
    title='Медианная цена квартир менее 50 м^2 в период с 2011 г. по 2014 г.',
    labels={'x': 'Месяц', 'y': 'Медианная цена, руб.'},
).show()
````
